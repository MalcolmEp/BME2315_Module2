## Disease Background: 

__Prevalence & incidence__

FSHD (Facioscapulohumeral Muscular Dystrophy) has an estimated prevalence of about 5 in 100,000. Although this disease affects both men and women, it has been shown to be more prevalent and severe in men. Overall there are no notable differences in prevalence relating to racial, geographic, or ethnic distribution, except for in Africa, where the prevalence seems to be less. However, this can be explained with the lack of ways to identify the disease in some sub-Sarahan parts of Africa. The estimated incidence is about 1 in 14,000 per year, however, as genetic testing and knowledge based around the disease has improved, the number is more likely closer to 1 in 7,500. (1)(2)


__Symptoms__

The main symptoms of FSHD (Facioscapulohumeral Muscular Dystrophy) all involve muscle weakening and deterioration. Usually showing in the teens or early twenties, muscle weakness begins in the face, shoulders, and upper arms asymmetrically. Weakness spreads slowly over a long period of time, 30+ years in some patients, and can reach disabling levels. About 20% of FSHD patients will need a wheelchair at some point in their life. Other signs of deterioration include trouble walking and closing eyes, joint pain, difficulty swallowing, hip and abdominal weakness, and winging scapula with a loss in range of motion. Hearing loss is also common at varying points in the progression of the disease. (3)


__Diagnosis__

If a patient suspects they are having early symptoms of FSHD, they will go to a doctor to see if they can diagnose it. Diagnosing FSHD starts with a basic physical exam with a doctor. This exam initially looks at the muscles affected. The doctor will then ask the patients about family and medical history, as FSHD is a hereditary condition. The doctor will then order tests for the patients. These tests include a blood test, a neurological test, a genetic test and a muscle biopsy. The blood test is done to measure the levels of serum creatine kinase in the blood. This is an enzyme that when the muscle fibers are deteriorating, is released into the bloodstream. It also looks for serum aldolase, which breaks down sugar into energy. Elevated levels of either enzyme can indicate muscle problems. The Neurological test is mainly used to rule out nervous system disorders, but also is used to detect muscle contractures. The muscle biopsy is used to detect abnormalities in muscle tissue. Finally, the genetic test is performed, to confirm a diagnosis of FSHD Type 1 or 2. (4)


__Standard of care treatments (& reimbursement)__

Because FSHD has no defined cure. Treatment usually revolves around symptom management, preserving muscle function, and maintaining quality of life. Low intensity exercise is recommended to preserve strength without damaging muscles further. This is often paired with physical/occupational therapy to keep the patient self-sufficient for longer, modifying work and living spaces as needed. Prosthetics and movement-aiding devices can be used for walking and standing up straight. If a patient is experiencing scapular winging but maintains strong deltoid muscles, scapular fixation surgery can be done, permanently anchoring the shoulder blade to the rib cage and allowing more movement of the arms. Standard commercial insurance and medicare cover most supportive and symptomatic care like therapy, pain medications, and genetic testing. However, there are many hidden and indirect costs that may result from a diagnosis, like home/vehicle modifications and loss of income. Many patients with FSHD are taken care of by unpaid family and friends which also is not accounted for in insurance or medicare/medicaid plans. (5)(6)


__Disease progression & prognosis__

The progression of FSHD is most commonly seen in the asymmetry of the body. The disease progresses down the body, starting at the face with difficulty whistling, drinking through a straw, or closing your eyes completely. It then moves to the shoulders, where it becomes hard to lift your arms above your head. Next is the upper arms, where lifting objects becomes overly difficult, however during this stage the forearms are usually not affected.  Finally, the lower body is affected, causing foot drop and the belly to protrude. This can cause a heavy fall risk for those affected by FSHD. The disease tends to begin in the teens or early 20s, where the progression on the muscles is typically slow. There are times where it can begin before the age of 10, and this form tends to progress much faster (7).


__Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)__

FSHD is caused by the transcription factor DUX4 gene in muscle cells, a toxic gene normally expressed in zygotic development but silenced for the rest of life. In the most common type 1 FSHD, a chromosomal region called D4Z4 isn’t repeated as many times as it should be. Healthy cells have anywhere from 11 to 150 repeats, while FSDH cells have less than 10. Missing the extra copies causes hypomethylation and relaxation of chromatin which partly allows the DUX4 gene encoded in the D4Z4 region to be transcribed. To be fully activated, the chromosome needs to be a specific allele with an activation site available to the DUX4 gene. If this is the case, FSHD is extremely likely to occur. Now that DUX4 is spread throughout the body, it causes muscle cell apoptosis and inhibits stem cell regeneration. Dying cells cause an inflammatory response, replacing lost muscle with fat and scar tissue which leads to loss of strength and range of motion over a patient’s life. (8)


__Clinical Trials/next-gen therapies__

A few current clinical trials and next-gen therapies include one from Soufflé Therapeutics which is a study being conducted to evaluate the safety, tolerability, and preliminary efficacy of SFL-0821 in adult patients with FSHD. SFL-0821 is an antibody that silences DUX4 expression in skeletal muscle. Another study is from FORTITUDE which is another study that is studying the safety of RNA therapy to block DUX4 (9).


## Data Set:

__The subject and visit/timepoint(s) analyzed__

34 FSHD subjects were recruited from multiple university sites and seen three times. One initial visit (time = 0), once at 12 months later, and once at 24 months later. However, almost none of the visits were performed at exactly 12 & 24 months. There is a notable lag before and after for most patients, likely due to scheduling/availability, and the pandemic occurring during the study.

__The tibialis anterior MRI data__

There are values for both left and right tibialis anterior, the shin muscle that lifts the foot. There are three MRI-derived measures per tibialis anterior. The three are fat fraction, a second fat measure, and a STIR measurement. There is a baseline value for each subject, 30 year-1 values and 25 year-2 values for fat fraction. STIR has 20 year-1 values.

__The water-image and fat-image files__

The water and fat image files are all individual MRI slices of the TA, one showing water content in the scan and the other showing fat content. Using computational analysis like pixel brightness and density, the images can be merged to identify the ratio of fat to water and get an overall fat content %.

__The tibialis anterior segmentation mask__

The mask is a 3D label marking which pixels belong to the tibialis anterior. It marks true or false, true if a pixel is part of the TA and false if it isn’t. It is useful in ignoring bones, other muscles, and the background when trying to analyze the data.


__The sampled locations along the muscle__

The muscles were sampled in 8 cross-sections from 10-80% along the length of each one. The slices aren’t even so they get plotted by distance in mm instead. The slices roughly start and end at 24 and 184mm respectively

__The meaning and units of the image signals__

For image signals such as raw water and fat, image intensities are in arbitrary scanner units, meaning only the ratio is meaningful. Fat fraction is the fat signal divided by the sum of fat and water signal and it is a percentage. Fvar is closely related, and measures fat fraction. Finally the STIR measurement which signals inflammation or active disease (most likely measured in % intensity).

__The clinical, biomarker, or functional measurements included in the dataset__

The full dataset includes genetics, mainly D4Z4 repeat amounts, biomarkers like fat percentage, DUX4 gene activity, and MRI fat fractions, strength of dorsal flexion, physical tests of function like walking and going up stairs, and a clinical severity score.

__The source of the data__

This data comes from Wong et al., Human Molecular Genetics, 2024. The patients were from three sites, University of Washington Medical Center, University of Rochester Medical Center, and Kansas University Medical Center.

The MRI data was scanned on 3T Siemens MRI scanners. Each patient had both legs scanned at markers placed at each subdivision so the same spot could be used for the biopsy. Water and fat show slightly differently on MRI scans and using this difference, separate images for each were able to be generated. Each slice is a cross-section of the leg and each pixel is about 1mm x 1mm. The mask was generated by an AI processor that segmented the TA on each scan, the scans were reviewed by a researcher afterwards for consistency and correctness. Each pixel gets a true or false value depending on whether it’s part of the TA or not. 

All of the information about the data came from the provided Wong et al. paper (10)

```python
# 1. import the packages and load the slice table

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("FSHD_TA_MRI_data_0020_TP1_TP2")
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")
manifest
```

Figure 1: This table has one row per MRI cross-section, 8 slices at each of the two visits. Each slice is labeled with its location in % along the TA muscle and the file names of the water and fat scan files. This is included because it is the base table for all of the data to be gathered from, interpreted, and used in computational methods. Noticeably, the slices are measured in distance from muscle start, this shows that the slices are not all the same distance apart even though they are the same thickness.

```python
# 2. Load the mask onto one slice and then apply it to all of them

row = manifest[manifest["timepoint"] == "TP1"].iloc[3]  # representative mid-muscle slice

water = np.load(DATA_DIR / row["water_file"]) # Reads each file as pixels
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool) # Sets each pixel to true or false

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
```

```python
# 3. Calculate fat fraction

denominator = water + fat
valid_pixels = denominator > 0

fat_fraction = np.divide(
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100

# Keep the whole supplied image region for the first analysis.
fat_fraction_full = fat_fraction.copy()

# Then apply the TA mask for the muscle-specific analysis.
fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_masked[~np.isnan(fat_fraction_masked)]

print(f"Whole supplied image region — mean: {np.mean(whole_image_values):.1f}%")
print(f"Whole supplied image region — minimum: {np.min(whole_image_values):.1f}%")
print(f"Whole supplied image region — maximum: {np.max(whole_image_values):.1f}%")
print()
print(f"TA muscle only — mean: {np.mean(muscle_values):.1f}%")
print(f"TA muscle only — minimum: {np.min(muscle_values):.1f}%")
print(f"TA muscle only — maximum: {np.max(muscle_values):.1f}%")
```

```python
# 4. Summarize fat-fraction map

# Find the bounding box of the TA mask
y_pixels, x_pixels = np.where(mask)

y0, y1 = y_pixels.min(), y_pixels.max() + 1
x0, x1 = x_pixels.min(), x_pixels.max() + 1

# Crop the TA-specific fat-fraction map
fat_fraction_TA = fat_fraction_masked[y0:y1, x0:x1]

fig, axes = plt.subplots(1, 4, figsize=(15, 4))

# Water image with TA mask outline
axes[0].imshow(water, cmap="gray")
axes[0].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[0].set_title("Water image + TA mask")

# Fat image with TA mask outline
axes[1].imshow(fat, cmap="gray")
axes[1].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[1].set_title("Fat image + TA mask")

# Fat fraction for the whole supplied image region
whole_image = axes[2].imshow(
    fat_fraction_full,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[2].contour(mask, levels=[0.5], colors="cyan", linewidths=2)
axes[2].set_title("Fat fraction: whole region")

# Zoomed-in fat fraction for the TA only
ta_image = axes[3].imshow(
    fat_fraction_TA,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[3].set_title("TA fat fraction")

for ax in axes:
    ax.axis("off")

# Both fat-fraction plots use the same 0–100% scale
fig.colorbar(
    ta_image,
    ax=axes[2:],
    fraction=0.046,
    pad=0.04,
    label="Fat fraction (%)"
)

fig.suptitle("Subject UW_0020, TP1, 40% along the tibialis anterior")
plt.show()
```

Figure 2: MRI images, TA mask, and fat-fraction maps for subject UW_0020, TP1, 40% along the tibialis anterior. This figure is included to show the raw outputs driving the analysis and to confirm that the map is correctly outlining the TA in each slice. The fat fraction: whole region section of the figure demonstrates why a TA mask is needed as is shows a lot of noise outside of the leg portion of the scan that would interfere with data collection.

```python
# 5. Create fat-fraction distribution histogram, whole image region vs TA muscle only

bins = np.linspace(0, 100, 31)
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

axes[0].hist(whole_image_values, bins=bins, color="tab:blue", alpha=0.7)
axes[0].set_title("A. Whole image region")
axes[0].set_xlabel("Fat fraction (%)")
axes[0].set_ylabel("Number of pixels")

axes[1].hist(muscle_values, bins=bins, color="tab:orange", alpha=0.8)
axes[1].axvline(np.mean(muscle_values), color="k", linestyle="--",
label=f"Mean = {np.mean(muscle_values):.1f}%")
axes[1].axvline(np.median(muscle_values), color="k", linestyle=":",
label=f"Median = {np.median(muscle_values):.1f}%")
axes[1].set_title("B. TA muscle only")
axes[1].set_xlabel("Fat fraction (%)")
axes[1].set_ylabel("Number of pixels")
axes[1].legend()

fig.suptitle("Fat-fraction distributions, UW_0020, TP1, 40% along the TA")
plt.tight_layout()
plt.show()
```

Figure 3: Fat fraction distributions in (A) the entire image region and (B) just the TA muscle. Mean and median lines are provided for the TA muscle percentages. This is included to show how fat is distributed across pixels in the scans as well as to compare the muscle to the whole image numbers. Notice that the whole image graph has large spikes at 0, 50, and 100%. These show that most pixels are either noise (50%), plain water (0%), or plain fat (100%). Looking at the TA muscle only shows why a mask is needed to pull out the useful data from the noise of the entire image.

```python
# 6. Calculate and show summary statistics

def calculate_fat_fraction(water, fat, mask):
    """Return a masked fat-fraction image in percent."""
    denominator = water + fat
    result = np.divide(
        fat,
        denominator,
        out=np.zeros_like(fat, dtype=float),
        where=denominator > 0
    ) * 100
    result[~mask] = np.nan
    return result

analysis_timepoint = "TP1"  # analyze one visit at a time
analysis_manifest = manifest[manifest["timepoint"] == analysis_timepoint]
results = []

for _, row in manifest.iterrows(): # Loops over both visits, TP1 and TP2
    water = np.load(DATA_DIR / row["water_file"])
    fat = np.load(DATA_DIR / row["fat_file"])
    mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
    ff = calculate_fat_fraction(water, fat, mask)

    results.append({
        "subject": row["subject"],
        "timepoint": row["timepoint"],
        "slice_id": row["slice_id"],
        "percent_along_muscle": row["percent_along_muscle"],
        "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
        "mean_fat_fraction_percent": np.nanmean(ff),
        "median_fat_fraction_percent": np.nanmedian(ff),
        "std_fat_fraction_percent": np.nanstd(ff),
        "n_muscle_pixels": int(mask.sum()),
    })

results_df = pd.DataFrame(results)
results_df
```

Figure 4: Fat-fraction summary statistics for each sampled location along the tibialis anterior. Each row is 1 MRI slice and its location, fat percentage, and number of fat pixels. It's included to show the application of the single-slice method from Figures 1–2 to every slice at both visits. At TP1, fat fraction varies greatly throughout the table with higher values near the start of the muscle and lower towards the middle. This shows that fat infiltration is concentrated more at one end of the muscle rather than spread evenly.

```python
# 7. Summarize measurements across muscle locations

plt.figure(figsize=(8, 5))
for timepoint, group in results_df.groupby("timepoint"):
    group = group.sort_values("distance_from_muscle_start_mm")
    plt.errorbar(
        group["distance_from_muscle_start_mm"],      # actual distance in mm
        group["mean_fat_fraction_percent"],
        yerr=group["std_fat_fraction_percent"],      # ±1 SD within each slice
        marker="o", capsize=4, label=timepoint
    )
plt.xlabel("Distance from muscle start (mm)")
plt.ylabel("Mean TA fat fraction (%)")
plt.title("Mean fat fraction along the tibialis anterior, UW_0020")
plt.ylim(0, 100)
plt.legend(title="Visit")
plt.grid(True, alpha=0.3)
plt.show()
```

Figure 5: Figure 3. Mean tibialis anterior fat fraction along the length of the muscle (mm). This is included to show whether fat infiltration is uniform along the muscle or not. The distribution seems to show that fat infiltration has high variation along the muscle, with large error bars indicating lots of variation in each cross-section.
