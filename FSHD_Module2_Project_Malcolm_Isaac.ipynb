## Project Goal:

FSHD is a slow progressing muscular dystrophy disease with no approved treatment and several experimental therapies currently being studied. Because the disease progresses so slowly, it can be difficult to identify an endpoint to look at as to whether a clinical trial is having a meaningful effect or not. Using natural-history data from 34 untreated FSHD patients (Wong et al., 2024) we will compare two measures of the tibialis anterior (TA), MRI fat fraction (a biomarker of muscle replacement by fat) and dorsiflexion strength (in kg).

Question: Which measure is more sensitive to change over approximately two years, TA fat fraction or dorsiflexion strength? We will assess sensitivity using the standardized response mean and the percentage of patients showing worsening. This will guide our decision as to which measure should be included as an exploratory endpoint in FSHD clinical trials.

## Disease Background: 

__Prevalence & incidence__

FSHD (Facioscapulohumeral Muscular Dystrophy) has an estimated prevalence of about 5 in 100,000. Although this disease affects both men and women, it has been shown to be more prevalent and severe in men. Overall there are no notable differences in prevalence relating to racial, geographic, or ethnic distribution, except for in Africa, where the prevalence seems to be less. However, this can be explained with the lack of ways to identify the disease in some sub-Sarahan parts of Africa. The estimated incidence is about 1 in 14,000 per year, however, as genetic testing and knowledge based around the disease has improved, the number is more likely closer to 1 in 7,500. (1)(2)


__Symptoms__

The main symptoms of FSHD (Facioscapulohumeral Muscular Dystrophy) all involve muscle weakening and deterioration. Usually showing in the teens or early twenties, muscle weakness begins in the face, shoulders, and upper arms asymmetrically. Weakness spreads slowly over a long period of time, 30+ years in some patients, and can reach disabling levels. About 20% of FSHD patients will need a wheelchair at some point in their life. Other signs of deterioration include trouble walking and closing eyes, joint pain, difficulty swallowing, hip and abdominal weakness, and winging scapula with a loss in range of motion. Hearing loss is also common at varying points in the progression of the disease. (3)


__Diagnosis__

If a patient suspects they are having early symptoms of FSHD, they will go to a doctor to see if they can diagnose it. Diagnosing FSHD starts with a basic physical exam with a doctor. This exam initially looks at the muscles affected. The doctor will then ask the patients about family and medical history, as FSHD is a hereditary condition. The doctor will then order tests for the patients. These tests include a blood test, a neurological test, a genetic test and a muscle biopsy. The blood test is done to measure the levels of serum creatine kinase in the blood. This is an enzyme that when the muscle fibers are deteriorating, is released into the bloodstream. It also looks for serum aldolase, which breaks down sugar into energy. Elevated levels of either enzyme can indicate muscle problems. The Neurological test is mainly used to rule out nervous system disorders, but also is used to detect muscle contractures. The muscle biopsy is used to detect abnormalities in muscle tissue. Finally, the genetic test is performed, to confirm a diagnosis of FSHD Type 1 or 2. (4)


__Standard of care treatments (& reimbursement)__

Because FSHD has no defined cure. Treatment usually revolves around symptom management, preserving muscle function, and maintaining quality of life. Low intensity exercise is recommended to preserve strength without damaging muscles further. This is often paired with physical/occupational therapy to keep the patient self-sufficient for longer, modifying work and living spaces as needed. Prosthetics and movement-aiding devices can be used for walking and standing up straight. If a patient is experiencing scapular winging but maintains strong deltoid muscles, scapular fixation surgery can be done, permanently anchoring the shoulder blade to the rib cage and allowing more movement of the arms. Standard commercial insurance and medicare cover most supportive and symptomatic care like therapy, pain medications, and genetic testing. However, there are many hidden and indirect costs that may result from a diagnosis, like home/vehicle modifications and loss of income. Many patients with FSHD are taken care of by unpaid family and friends which also is not accounted for in insurance or medicare/medicaid plans. (5)(6)


__Disease progression & prognosis__

The progression of FSHD is most commonly seen in the asymmetry of the body. The disease progresses down the body, starting at the face with difficulty whistling, drinking through a straw, or closing your eyes completely. It then moves to the shoulders, where it becomes hard to lift your arms above your head. Next is the upper arms, where lifting objects becomes overly difficult, however during this stage the forearms are usually not affected.  Finally, the lower body is affected, causing foot drop and the belly to protrude. This can cause a heavy fall risk for those affected by FSHD. The disease tends to begin in the teens or early 20s, where the progression on the muscles is typically slow. There are times where it can begin before the age of 10, and this form tends to progress much faster (7).


__Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)__

FSHD is caused by the transcription factor DUX4 gene in muscle cells, a toxic gene normally expressed in zygotic development but silenced for the rest of life. In the most common type 1 FSHD, a chromosomal region called D4Z4 isn’t repeated as many times as it should be. Healthy cells have anywhere from 11 to 150 repeats, while FSDH cells have less than 10. Missing the extra copies causes hypomethylation and relaxation of chromatin which partly allows the DUX4 gene encoded in the D4Z4 region to be transcribed. To be fully activated, the chromosome needs to be a specific allele with an activation site available to the DUX4 gene. If this is the case, FSHD is extremely likely to occur. Now that DUX4 is spread throughout the body, it causes muscle cell apoptosis and inhibits stem cell regeneration. Dying cells cause an inflammatory response, replacing lost muscle with fat and scar tissue which leads to loss of strength and range of motion over a patient’s life. (8)


__Clinical Trials/next-gen therapies__

A few current clinical trials and next-gen therapies include one from Soufflé Therapeutics which is a study being conducted to evaluate the safety, tolerability, and preliminary efficacy of SFL-0821 in adult patients with FSHD. SFL-0821 is an antibody that silences DUX4 expression in skeletal muscle. Another study is from FORTITUDE which is another study that is studying the safety of RNA therapy to block DUX4 (9).


## Data Set:

__The subject and visit/timepoint(s) analyzed__

34 FSHD subjects were recruited from multiple university sites and seen three times. One initial visit (time = 0), once at 12 months later, and once at 24 months later. However, almost none of the visits were performed at exactly 12 & 24 months. There is a notable lag before and after for most patients, likely due to scheduling/availability, and the pandemic occurring during the study.

__The tibialis anterior MRI data__

There are values for both left and right tibialis anterior, the shin muscle that lifts the foot. There are three MRI-derived measures per tibialis anterior. The three are fat fraction, a second fat measure, and a STIR measurement. There is a baseline value for each subject, 30 year-1 values and 25 year-2 values for fat fraction. STIR has 20 year-1 values.

__The water-image and fat-image files__

The water and fat image files are all individual MRI slices of the TA, one showing water content in the scan and the other showing fat content. Using computational analysis like pixel brightness and density, the images can be merged to identify the ratio of fat to water and get an overall fat content %.

__The tibialis anterior segmentation mask__

The mask is a 3D label marking which pixels belong to the tibialis anterior. It marks true or false, true if a pixel is part of the TA and false if it isn’t. It is useful in ignoring bones, other muscles, and the background when trying to analyze the data.


__The sampled locations along the muscle__

The muscles were sampled in 8 cross-sections from 10-80% along the length of each one. The slices aren’t even so they get plotted by distance in mm instead. The slices roughly start and end at 24 and 184mm respectively

__The meaning and units of the image signals__

For image signals such as raw water and fat, image intensities are in arbitrary scanner units, meaning only the ratio is meaningful. Fat fraction is the fat signal divided by the sum of fat and water signal and it is a percentage. Fvar is closely related, and measures fat fraction. Finally the STIR measurement which signals inflammation or active disease (most likely measured in % intensity).

__The clinical, biomarker, or functional measurements included in the dataset__

The full dataset includes genetics, mainly D4Z4 repeat amounts, biomarkers like fat percentage, DUX4 gene activity, and MRI fat fractions, strength of dorsal flexion, physical tests of function like walking and going up stairs, and a clinical severity score.

__The source of the data__

This data comes from Wong et al., Human Molecular Genetics, 2024. The patients were from three sites, University of Washington Medical Center, University of Rochester Medical Center, and Kansas University Medical Center.

The MRI data was scanned on 3T Siemens MRI scanners. Each patient had both legs scanned at markers placed at each subdivision so the same spot could be used for the biopsy. Water and fat show slightly differently on MRI scans and using this difference, separate images for each were able to be generated. Each slice is a cross-section of the leg and each pixel is about 1mm x 1mm. The mask was generated by an AI processor that segmented the TA on each scan, the scans were reviewed by a researcher afterwards for consistency and correctness. Each pixel gets a true or false value depending on whether it’s part of the TA or not. 

All of the information about the data came from the provided Wong et al. paper (10)


## Interpolation Analysis:

Interpolation is done because not all data points were collected at the same time. Many patients visited eariler or later than 12 and 24 months. Assuming all data points happen at 12 and 24 months would mean extrapolating patients who never reached 12 or 24 months. This would lead to inaccurace or untrustworthy results. Instead, what can be done is find a formula that plots what each data point WOULD be at the earliest time a data point was collected, that way, no data is being extrapolated at a time after it was actually collected. 

To do this, the earliest point of data collection in year 1 and year 2 is found (11.15 months and 22.13 months) and a quadratic curve is calculated that exactly fits the given data points for each patient. In addition, the first two data points are seperately modeled by a linear exact fit.

The inputs for the new data are each patient's three visits, all at varying times of the 1 and 2 year period, and the outputs are the interpolated data points at T = 11.15 and 22.13 months. These new interpolated data points for fat fraction and dorsiflexion strength are saved in a pandas dataframe to be used for data analysis without any extrapolation. It is important to note that these are technically not real collected data points, no patient except the earliest one is actually measured at the date being interpolated.

Some limitations for this interpolation are sample size and null data points. Only three visits means a larger variability between patients and more uncertain interpolation results. This can make interpolation less accurate as the estimates need to do more with less data to work with. Some patients also have no first or second visit which meant some data was left out of the sample. This not only reduces the patient sample size, but can bias the results against factors that might be causing patients to miss visits.

```python
# Import relevant libraries and fetch the earliest year 1 and 2 visits

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

df = pd.read_csv("Module2_FSHD_Dataset.csv")
T1 = df["lag_months_y1"].min()   # earliest Year-1 visit 
T2 = df["lag_months_y2"].min()   # earliest Year-2 visit
print(f"T1 = {T1:.2f} months, T2 = {T2:.2f} months")
```

```python
# Pull out one patient's values and make an array of fat fraction and dorsiflexion strength at each time point

def get_ff(row, side):
    return np.array([row[f"ff_baseline_{side}"], # side is either R or L and is accepted as a parameter to the function
                     row[f"ff_y1_{side}"],
                     row[f"ff_y2_{side}"]])

def get_dorsi(row, side):
    return np.array([row[f"dorsiflexion_qmt_{side}_baseline"],
                     row[f"dorsiflexion_qmt_{side}_y1"],
                     row[f"dorsiflexion_qmt_{side}_y2"]])
```

```python
# Define interpolation functions for linear and quadratic

def interpolate_subject(lag_months, values):
    # Linear estimate at T1, baseline to year 1
    if np.isnan(lag_months[:2]).any() or np.isnan(values[:2]).any():
        est_T1 = np.nan
    else:
        f_linear = interp1d(lag_months[:2], values[:2], kind="linear")
        est_T1 = float(f_linear(T1))

    # Quadratic estimate at T2, all three visits
    # Double check that the year 1 visit is before the year 2 visit because patient 0031 has the same number for both and interp1d can't fit a curve through two points at the same x value
    if np.isnan(lag_months).any() or np.isnan(values).any() or lag_months[1] >= lag_months[2]:
        est_T2 = np.nan
    else:
        f_quad = interp1d(lag_months, values, kind="quadratic")
        est_T2 = float(f_quad(T2))

    return est_T1, est_T2
```

```python
# Loop interpolation function for every patient and both legs

results = []

for idx, row in df.iterrows():
    lag_months = np.array([row["lag_months_baseline"],
                           row["lag_months_y1"],
                           row["lag_months_y2"]])

    for side in ["R", "L"]:
        ff = get_ff(row, side)
        dorsi = get_dorsi(row, side)

        ff_T1, ff_T2 = interpolate_subject(lag_months, ff)
        dorsi_T1, dorsi_T2 = interpolate_subject(lag_months, dorsi)

        results.append({
            "subject_id": row["subject_id"], "side": side,
            "ff_baseline": ff[0],       "ff_T1": ff_T1,       "ff_T2": ff_T2,
            "dorsi_baseline": dorsi[0], "dorsi_T1": dorsi_T1, "dorsi_T2": dorsi_T2,
        })

interp_df = pd.DataFrame(results)
interp_df.to_csv("interpolated_estimates.csv", index=False)
print(interp_df.notna().sum())   # how many legs have a value in each column
interp_df.head()
```

```python
# Plot measured values vs estimates

row = df.loc[df["subject_id"] == "13-0004"].iloc[0]
lag_months = np.array([row["lag_months_baseline"], row["lag_months_y1"], row["lag_months_y2"]])

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
for ax, values, label in [(axes[0], get_ff(row, "R"), "Fat fraction (%)"),
                          (axes[1], get_dorsi(row, "R"), "Dorsiflexion strength (kg)")]:
    est_T1, est_T2 = interpolate_subject(lag_months, values)
    t_curve = np.linspace(0, lag_months[-1], 200)
    ax.plot(t_curve, interp1d(lag_months, values, kind="quadratic")(t_curve), "--", label="Quadratic curve")
    ax.plot(lag_months[:2], values[:2], "-", color="tab:pink", label="Linear (baseline → Year 1)")
    ax.scatter(lag_months, values, s=70, label="Measured visits")
    ax.scatter([T1, T2], [est_T1, est_T2], marker="D", s=100, color="red", label="Estimates at T1, T2")
    ax.set_xlabel("Months since baseline")
    ax.set_ylabel(label)
    ax.legend()
    ax.grid(alpha=0.3)
fig.suptitle("Subject 13-0004, right TA: interpolated and original data at common time points")
plt.tight_layout()
plt.show()
```

This graph shows the interpolated and original data for fat fraction and dorsiflexion strength for a patient with a particularly late T2 visit. The new data points are all at at the earliest T1 and T2 as shown in the graph 

Each curve assumes a slihtly different trajectory and resulting data point for the patient values. Linear assumes a linear increase based off of the starting point and T1, while quadratic assimes a quadratic trajectory and resulting data point based off of T0, T1, and T2. Notable, this patient's strength went UP even though FSHD is progressive, this may point to dorsiflexion not being a very consistent measire, or in other words, having large variance between patients.

## MRI Data Analysis Methodology:

```python
# 1. import remaining packages and load the slice table

from pathlib import Path

DATA_DIR = Path("FSHD_TA_MRI_data_0020_TP1_TP2")
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")
manifest
```

Figure 1: This table has one row per MRI cross-section, 8 slices at each of the two visits. Each slice is labeled with its location in % along the TA muscle and the file names of the water and fat scan files. This is included because it is the base table for all of the data to be gathered from, interpreted, and used in computational methods. Noticeably, the slices are measured in distance from muscle start, this shows that the slices are not all the same distance apart even though they are the same thickness.

```python
# 2. Load the mask onto one slice and then apply it to all of them

row = manifest[manifest["timepoint"] == "TP1"].iloc[3]  # representative mid-muscle slice

water = np.load(DATA_DIR / row["water_file"]) # Reads each file as pixels
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool) # Sets each pixel to true or false

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
```

```python
# 3. Calculate fat fraction

denominator = water + fat
valid_pixels = denominator > 0

fat_fraction = np.divide(
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100

# Keep the whole supplied image region for the first analysis.
fat_fraction_full = fat_fraction.copy()

# Then apply the TA mask for the muscle-specific analysis.
fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_masked[~np.isnan(fat_fraction_masked)]

print(f"Whole supplied image region — mean: {np.mean(whole_image_values):.1f}%")
print(f"Whole supplied image region — minimum: {np.min(whole_image_values):.1f}%")
print(f"Whole supplied image region — maximum: {np.max(whole_image_values):.1f}%")
print()
print(f"TA muscle only — mean: {np.mean(muscle_values):.1f}%")
print(f"TA muscle only — minimum: {np.min(muscle_values):.1f}%")
print(f"TA muscle only — maximum: {np.max(muscle_values):.1f}%")
```

```python
# 4. Summarize fat-fraction map

# Find the bounding box of the TA mask
y_pixels, x_pixels = np.where(mask)

y0, y1 = y_pixels.min(), y_pixels.max() + 1
x0, x1 = x_pixels.min(), x_pixels.max() + 1

# Crop the TA-specific fat-fraction map
fat_fraction_TA = fat_fraction_masked[y0:y1, x0:x1]

fig, axes = plt.subplots(1, 4, figsize=(15, 4))

# Water image with TA mask outline
axes[0].imshow(water, cmap="gray")
axes[0].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[0].set_title("Water image + TA mask")

# Fat image with TA mask outline
axes[1].imshow(fat, cmap="gray")
axes[1].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[1].set_title("Fat image + TA mask")

# Fat fraction for the whole supplied image region
whole_image = axes[2].imshow(
    fat_fraction_full,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[2].contour(mask, levels=[0.5], colors="cyan", linewidths=2)
axes[2].set_title("Fat fraction: whole region")

# Zoomed-in fat fraction for the TA only
ta_image = axes[3].imshow(
    fat_fraction_TA,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[3].set_title("TA fat fraction")

for ax in axes:
    ax.axis("off")

# Both fat-fraction plots use the same 0–100% scale
fig.colorbar(
    ta_image,
    ax=axes[2:],
    fraction=0.046,
    pad=0.04,
    label="Fat fraction (%)"
)

fig.suptitle("Subject UW_0020, TP1, 40% along the tibialis anterior")
plt.show()
```

```python
# 6. Calculate and show summary statistics

def calculate_fat_fraction(water, fat, mask):
    """Return a masked fat-fraction image in percent."""
    denominator = water + fat
    result = np.divide(
        fat,
        denominator,
        out=np.zeros_like(fat, dtype=float),
        where=denominator > 0
    ) * 100
    result[~mask] = np.nan
    return result

analysis_timepoint = "TP1"  # analyze one visit at a time
analysis_manifest = manifest[manifest["timepoint"] == analysis_timepoint]
results = []

for _, row in manifest.iterrows(): # Loops over both visits, TP1 and TP2
    water = np.load(DATA_DIR / row["water_file"])
    fat = np.load(DATA_DIR / row["fat_file"])
    mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
    ff = calculate_fat_fraction(water, fat, mask)

    results.append({
        "subject": row["subject"],
        "timepoint": row["timepoint"],
        "slice_id": row["slice_id"],
        "percent_along_muscle": row["percent_along_muscle"],
        "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
        "mean_fat_fraction_percent": np.nanmean(ff),
        "median_fat_fraction_percent": np.nanmedian(ff),
        "std_fat_fraction_percent": np.nanstd(ff),
        "n_muscle_pixels": int(mask.sum()),
    })

results_df = pd.DataFrame(results)
results_df
```

## Sensitivity to Change Analysis: Fat Fraction vs Dorsiflexion Strength

This coding portion will statistically analyze the sensitivity of fat fraction and dorsiflexion strength over time.

```python
# First, the sign of dorsiflexsion strength needs to be flipped to make positive indicate a worsening condition for both measures

analysis_df = interp_df.copy()   # copy so interp_df isn't accidentally changed

analysis_df["ff_change"] = analysis_df["ff_T2"] - analysis_df["ff_baseline"]
analysis_df["dorsi_change"] = analysis_df["dorsi_T2"] - analysis_df["dorsi_baseline"]
analysis_df["ff_worsening"] = analysis_df["ff_change"]
analysis_df["dorsi_worsening"] = -analysis_df["dorsi_change"]
```

```python
# Second, only legs with data values for both measures are kept so measures are always compared on the same people

matched = analysis_df.dropna(subset=["ff_worsening", "dorsi_worsening"])
print(f"Legs with both measures: {len(matched)} from {matched['subject_id'].nunique()} patients")
```

```python
# Define standardized response mean, percent of patients worsening, and bootstrap at 95% confidence. (Bootstrapping is a way to find out how much a result could change by complete chance)

def srm(changes): # SRM = mean change / standard deviation of change
    return np.mean(changes) / np.std(changes, ddof=1)

def pct_worsening(changes): # Percent of legs that got worse (when worsening is > 0)
    return 100 * np.mean(changes > 0)

def bootstrap_srm_ci(data, column, n_boot=5000, seed=2315): # 95% confidence interval for SRM by resampling patients with replacement
    rng = np.random.default_rng(seed)
    subjects = data["subject_id"].unique() # store each patient's values so both legs of one person stay together
    values_by_subject = {s: data.loc[data["subject_id"] == s, column].values for s in subjects}
    boot_srms = []
    for i in range(n_boot):
        picked = rng.choice(subjects, size=len(subjects), replace=True)
        sample = np.concatenate([values_by_subject[s] for s in picked])
        boot_srms.append(srm(sample))
    return np.percentile(boot_srms, [2.5, 97.5])

summary_rows = []
for measure, column in [("Fat fraction", "ff_worsening"),
                        ("Dorsiflexion strength", "dorsi_worsening")]:
    for leg in ["R", "L", "Both"]:
        if leg == "Both":
            data = matched
        else:
            data = matched[matched["side"] == leg]
        ci_low, ci_high = bootstrap_srm_ci(data, column)
        summary_rows.append({
            "measure": measure, "leg": leg, "n_legs": len(data), # Creating the chart with row titles and data points
            "mean_worsening": data[column].mean(),
            "sd_worsening": data[column].std(),
            "SRM": srm(data[column]),
            "CI_low": ci_low, "CI_high": ci_high,
            "pct_worsening": pct_worsening(data[column]),
        })

summary_df = pd.DataFrame(summary_rows)
summary_df.round(2)
```

This chart mainly shows the mean, standard deviation, and percent worsening for fat fraction and dorsiflexion strength for the left leg, right leg, and both. This is included to extract the data values all at once and compare the percent worsening for each measure to each other. Notice that the percent worsening for fat fraction is over 80 percent with a 95% confidence interval low above zero, this indicates a statistically significant and consistent increase in fat fraction over time in these patients. Dorsiflexion strength on the other hand, has a percent worsening below 50% and a confidence interval low below zero which indicates that statistically, the data presented is no more than the result of random noise.

```python
# Plot the standardized response means of fat fraction anddorsiflexion strength

fig, ax = plt.subplots(figsize=(8, 5))
legs = ["R", "L", "Both"]
x = np.arange(len(legs))   # bar positions at 0, 1, and 2
width = 0.35

for i, (measure, color) in enumerate([("Fat fraction", "tab:orange"),
                                      ("Dorsiflexion strength", "tab:blue")]):
    rows = summary_df[summary_df["measure"] == measure]
    error = [rows["SRM"] - rows["CI_low"], rows["CI_high"] - rows["SRM"]]   # distance down & distance up
    ax.bar(x + (i - 0.5) * width, rows["SRM"], width, yerr=error, capsize=5,
           color=color, label=measure)

ax.axhline(0, color="black", linewidth=1) # Creating graph with labels, colors, and error bars
for level in [0.2, 0.5, 0.8]:   
    ax.axhline(level, color="gray", linestyle=":", linewidth=1)
ax.set_xticks(x)
ax.set_xticklabels(["Right TA", "Left TA", "Both legs"])
ax.set_ylabel("Standardized response mean (positive = worsening)")
ax.set_title("Sensitivity to change from baseline to T2 (22.13 months)")
ax.legend(loc="lower left")
ax.grid(axis="y", alpha=0.3)
plt.show()
```

Shown here is the mean percent worsening for left, right, and both legs with error bars of fat fraction and dorsiflexion. This gives a good visualization of the error bars not reaching zero for fat fraction, but all crossing zero for dorsiflexion strength, which indicates no startstically signifigant change over time.

```python
# Waterfall plot for % worsening

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
for ax, column, title, ylabel in [
        (axes[0], "ff_worsening", "Fat fraction", "Increase in fat fraction (% points)"), # Creating bar chart with colors, labels, and data
        (axes[1], "dorsi_worsening", "Dorsiflexion strength", "Loss of strength (kg)")]:
    values = matched[column].sort_values(ascending=False).values
    colors = ["tab:red" if v > 0 else "tab:gray" for v in values]
    ax.bar(range(len(values)), values, color=colors)
    ax.axhline(0, color="black", linewidth=1) 
    ax.set_title(f"{title}: {pct_worsening(values):.0f}% of legs worsened")
    ax.set_xlabel("Individual legs (sorted)")
    ax.set_ylabel(ylabel)
    ax.grid(axis="y", alpha=0.3)
fig.suptitle("Change from baseline to T2 for each leg (red = worse)")
plt.tight_layout()
plt.show()
```

Shown here is a waterfall plot of each measure. A waterfall plot graphs each value of fat fraction or dorsiflexion strength for each patient in descending order. This is useful to visualize the fraction of patients that worsened vs got better over time. A completely random sample would have a worsening rate of approximately 50%, so the dorsiflexion worsening rate of 43% is basically equivalent to random noise. The 81% worsening rate for fat fraction is much better and shows that a majority of patients' fat fraction gets worse over time.

## Conclusions:

According to the statistical tests and visualizations, it is clear that fat fraction is a far better biomarker for worsening condition over time than dorsiflexion strength. It worsened much more often (81% compared to 43%) and most importantly, had a 95% confidence interval for percent change very far from 0 which is a statistically significant change over time. This seems to indicate that fat fraction could be very useful as a biomarker for clinical trials as it changes consistantly over time, is the root cause for muscle weakness and fatigue, and isn't as susceptible to confounding factors of the individual patient, test, and day performed, like a walking test would be. Still, a larger sample size would most likely be needed to confirm fat fraction as a clinically viable biomarker as a sample size of 20-30 patients just isn't very large and could still be very biased or exhibiting a trend that is only visible on a small scale. If a much larger sample size of patients was used and a statisically significant change in fat fraction over time was still observed, its use as a biomarker in clinical trials could be much more viable. Because this data is aquired by an MRI scan rather than a physical test or biopsy, this is a particularly non-invasive biomarker to collect and probably quite accurate. The only issue with MRI imaging is that smaller or more rural hospitals may not have MRI machines to do this kind of analysis, requiring more invasive or tiring procedures.
